```python
#Import pandas as PD 
import pandas as pd 
import numpy as np
```

```python
#Reading all 6 Data files 
Netflix_stock_action= pd.read_csv("Netflix_stock_action.csv")
Netflix_stock_dividends= pd.read_csv("Netflix_stock_dividends.csv")
Netflix_stock_history= pd.read_csv("Netflix_stock_history.csv")
Netflix_stock_spilts= pd.read_csv("Netflix_stock_spilts.csv")
Netflix_stock_info=pd.read_csv("Netflix_stock_info.csv")
```

```python
#Check All Files

print("ACTION")
print(Netflix_stock_action.shape)
print(Netflix_stock_action.head())

print("DIVIDENDS")
print(Netflix_stock_dividends.shape)
print(Netflix_stock_dividends.head())

print("HISTORY")
print(Netflix_stock_history.shape)
print(Netflix_stock_history.head())

print("SPILTS")
print(Netflix_stock_spilts.shape)
print(Netflix_stock_spilts.head())

print("INFO")
print(Netflix_stock_info.shape)
print(Netflix_stock_info.head())

#4. Clean Netflix_stock_history

Netflix_stock_history["Date"] = pd.to_datetime(
    Netflix_stock_history["Date"],
    errors="coerce"
)

numeric_cols = [
    "Open",
    "High",
    "Low",
    "Close",
    "Adj Close",
    "Volume"
]

for col in numeric_cols:
    Netflix_stock_history[col] = pd.to_numeric(
        Netflix_stock_history[col],
        errors="coerce"
    )

Netflix_stock_history = (
    Netflix_stock_history
    .drop_duplicates(subset="Date")
    .sort_values("Date")
    .reset_index(drop=True)
)

print(Netflix_stock_history.info())
print(Netflix_stock_history.isna().sum())

#5. Check Negative Values

print(
    Netflix_stock_history[
        numeric_cols
    ].lt(0).sum()
)

#6. Check OHLC Logic

bad_rows = Netflix_stock_history[
    (Netflix_stock_history["High"] <
     Netflix_stock_history["Low"]) |

    (Netflix_stock_history["High"] <
     Netflix_stock_history["Open"]) |

    (Netflix_stock_history["High"] <
     Netflix_stock_history["Close"]) |

    (Netflix_stock_history["Low"] >
     Netflix_stock_history["Open"]) |

    (Netflix_stock_history["Low"] >
     Netflix_stock_history["Close"])
]

print("Invalid OHLC rows:", len(bad_rows))
print(bad_rows)

#7. Handle Missing History Rows

# Handle Missing History Rows
Netflix_stock_history = Netflix_stock_history.dropna(
    subset=[
        "Date",
        "Open",
        "High",
        "Low",
        "Close",
        "Adj Close",
        "Volume"
    ]
)

print(
    Netflix_stock_history.isna().sum()
)

#For stock prices, do not automatically replace missing prices with the mean. Missing observations should first be investigated because stock prices are time-series data.

#8. Clean Netflix_stock_spilts

Netflix_stock_spilts["Date"] = pd.to_datetime(
    Netflix_stock_spilts["Date"],
    errors="coerce"
)

Netflix_stock_spilts["Stock Splits"] = pd.to_numeric(
    Netflix_stock_spilts["Stock Splits"],
    errors="coerce"
)

Netflix_stock_spilts = (
    Netflix_stock_spilts
    .dropna()
    .drop_duplicates(subset="Date")
    .sort_values("Date")
)

print(Netflix_stock_spilts)

#9. Clean Netflix_stock_action

Netflix_stock_action["Date"] = pd.to_datetime(
    Netflix_stock_action["Date"],
    errors="coerce"
)

Netflix_stock_action["Dividends"] = pd.to_numeric(
    Netflix_stock_action["Dividends"],
    errors="coerce"
)

Netflix_stock_action["Stock Splits"] = pd.to_numeric(
    Netflix_stock_action["Stock Splits"],
    errors="coerce"
)

Netflix_stock_action = (
    Netflix_stock_action
    .drop_duplicates(subset="Date")
    .sort_values("Date")
)

print(Netflix_stock_action)

#10. Clean Netflix_stock_dividends

Netflix_stock_dividends["Date"] = pd.to_datetime(
    Netflix_stock_dividends["Date"],
    errors="coerce"
)

Netflix_stock_dividends["Dividends"] = pd.to_numeric(
    Netflix_stock_dividends["Dividends"],
    errors="coerce"
)

print(Netflix_stock_dividends.shape)
print(Netflix_stock_dividends)
```

```python
#Clean Netflix_stock_history

Netflix_stock_history["Date"] = pd.to_datetime(
    Netflix_stock_history["Date"],
    errors="coerce"
)

numeric_cols = [
    "Open",
    "High",
    "Low",
    "Close",
    "Adj Close",
    "Volume"
]

for col in numeric_cols:
    Netflix_stock_history[col] = pd.to_numeric(
        Netflix_stock_history[col],
        errors="coerce"
    )

Netflix_stock_history = (
    Netflix_stock_history
    .drop_duplicates(subset="Date")
    .sort_values("Date")
    .reset_index(drop=True)
)

print(Netflix_stock_history.info())
print(Netflix_stock_history.isna().sum())

#Check Negative Values

print(
    Netflix_stock_history[
        numeric_cols
    ].lt(0).sum()
)

#Check OHLC Logic

bad_rows = Netflix_stock_history[
    (Netflix_stock_history["High"] <
     Netflix_stock_history["Low"]) |

    (Netflix_stock_history["High"] <
     Netflix_stock_history["Open"]) |

    (Netflix_stock_history["High"] <
     Netflix_stock_history["Close"]) |

    (Netflix_stock_history["Low"] >
     Netflix_stock_history["Open"]) |

    (Netflix_stock_history["Low"] >
     Netflix_stock_history["Close"])
]

print("Invalid OHLC rows:", len(bad_rows))
print(bad_rows)

#Handle Missing History Rows

Netflix_stock_history = Netflix_stock_history.dropna(
    subset=[
        "Date",
        "Open",
        "High",
        "Low",
        "Close",
        "Adj Close",
        "Volume"
    ]
)

print(
    Netflix_stock_history.isna().sum()
)

#Clean Netflix_stock_spilts

Netflix_stock_spilts["Date"] = pd.to_datetime(
    Netflix_stock_spilts["Date"],
    errors="coerce"
)

Netflix_stock_spilts["Stock Splits"] = pd.to_numeric(
    Netflix_stock_spilts["Stock Splits"],
    errors="coerce"
)

Netflix_stock_spilts = (
    Netflix_stock_spilts
    .dropna()
    .drop_duplicates(subset="Date")
    .sort_values("Date")
)

print(Netflix_stock_spilts)

#9Clean Netflix_stock_action

Netflix_stock_action["Date"] = pd.to_datetime(
    Netflix_stock_action["Date"],
    errors="coerce"
)

Netflix_stock_action["Dividends"] = pd.to_numeric(
    Netflix_stock_action["Dividends"],
    errors="coerce"
)

Netflix_stock_action["Stock Splits"] = pd.to_numeric(
    Netflix_stock_action["Stock Splits"],
    errors="coerce"
)

Netflix_stock_action = (
    Netflix_stock_action
    .drop_duplicates(subset="Date")
    .sort_values("Date")
)

print(Netflix_stock_action)

#Clean Netflix_stock_dividends

Netflix_stock_dividends["Date"] = pd.to_datetime(
    Netflix_stock_dividends["Date"],
    errors="coerce"
)

Netflix_stock_dividends["Dividends"] = pd.to_numeric(
    Netflix_stock_dividends["Dividends"],
    errors="coerce"
)

print(Netflix_stock_dividends.shape)
print(Netflix_stock_dividends)


#Clean Netflix_stock_info

print(Netflix_stock_info.head())
print(Netflix_stock_info.info())

#Create New Analysis Variables

Netflix_stock_history["Daily_Return"] = (
    Netflix_stock_history["Adj Close"].pct_change()
    * 100
)

Netflix_stock_history["Daily_Range"] = (
    Netflix_stock_history["High"]
    - Netflix_stock_history["Low"]
)

Netflix_stock_history["Open_Close_Change"] = (
    Netflix_stock_history["Close"]
    - Netflix_stock_history["Open"]
)

Netflix_stock_history["Year"] = (
    Netflix_stock_history["Date"].dt.year
)

Netflix_stock_history["Month"] = (
    Netflix_stock_history["Date"].dt.month
)

print(Netflix_stock_history.head())

#Flag Outliers Instead of Automatically Deleting Them

Q1 = Netflix_stock_history["Daily_Return"].quantile(0.25)
Q3 = Netflix_stock_history["Daily_Return"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

Netflix_stock_history["Return_Outlier"] = (
    (Netflix_stock_history["Daily_Return"] < lower)
    |
    (Netflix_stock_history["Daily_Return"] > upper)
)

print(
    Netflix_stock_history["Return_Outlier"]
    .value_counts()
)

#Save Cleaned Files

Netflix_stock_history.to_csv(
    "Netflix_stock_history_clean.csv",
    index=False
)

Netflix_stock_action.to_csv(
    "Netflix_stock_action_clean.csv",
    index=False
)

Netflix_stock_dividends.to_csv(
    "Netflix_stock_dividends_clean.csv",
    index=False
)

Netflix_stock_spilts.to_csv(
    "Netflix_stock_spilts_clean.csv",
    index=False
)

Netflix_stock_info.to_csv(
    "Netflix_stock_info_clean.csv",
    index=False
)

print("All files cleaned and saved.")
```
